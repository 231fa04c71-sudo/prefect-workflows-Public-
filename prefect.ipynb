```python
from prefect import flow, task

@task
def collect_order():
    print("Order received")
    return "Pizza Order"

@task
def prepare_order(order):
    print("Preparing:", order)
    return "Pizza Ready"

@task
def deliver_order(order):
    print("Delivering:", order)

@flow(name="Restaurant Workflow", log_prints=True)
def restaurant_workflow():
    order = collect_order()
    prepared_order = prepare_order(order)
    deliver_order(prepared_order)

restaurant_workflow()
```

```python
from prefect import flow, task

@task
def load_data():
    print("STEP 1: Loading data")
    return [10, 20, 30, 40]

@task
def preprocess_data(data):
    print("STEP 2: Preprocessing data")
    return [x / 10 for x in data]

@task
def train_model(data):
    print("STEP 3: Training model")
    print("Training data:", data)
    return "trained_model"

@task
def evaluate_model(model):
    print("STEP 4: Evaluating model")
    print("Model:", model)
    print("Accuracy = 0.90")
    return 0.90

@flow(name="Simple ML Workflow", log_prints=True)
def simple_ml_workflow():
    data = load_data()
    processed_data = preprocess_data(data)
    model = train_model(processed_data)
    evaluate_model(model)

simple_ml_workflow()
```

```python
from prefect import flow, task

attempt = 0

@task(retries=2, retry_delay_seconds=2)
def connect_database():
    global attempt
    attempt += 1

    print("Connection Attempt:", attempt)

    if attempt == 1:
        raise Exception("Database temporarily unavailable")

    print("Database connected")
    return "Hospital Data"


@task
def preprocess_data(data):
    print("Preprocessing:", data)


@flow(name="Hospital Workflow", log_prints=True)
def hospital_workflow():
    data = connect_database()
    preprocess_data(data)


hospital_workflow()
```

```python
from prefect import flow, task

attempt_counter = {"count": 0}

@task(retries=2, retry_delay_seconds=2)
def connect_database():
    attempt_counter["count"] += 1

    print("Connection Attempt:", attempt_counter["count"])

    if attempt_counter["count"] == 1:
        raise Exception("Database temporarily unavailable")

    return "Hospital Data"


@task
def preprocess_data(data):
    print("Preprocessing:", data)


@flow(name="Retry Demo", log_prints=True)
def hospital_workflow():
    data = connect_database()
    preprocess_data(data)


hospital_workflow()
```

```python
from prefect import flow, task

@task
def train_model():
    return "model"

@task
def evaluate_model(model):
    return 0.87

@task
def save_model(model):
    print("Model saved")

@flow(name="Model Acceptance Flow", log_prints=True)
def ml_workflow():
    model = train_model()
    accuracy = evaluate_model(model)

    if accuracy >= 0.85:
        save_model(model)
        print("Model accepted")
    else:
        print("Model rejected")

ml_workflow()
```
